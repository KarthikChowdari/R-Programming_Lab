```r
# Remove duplicate records

transactions_data <- transactions_data %>%
  distinct()

products_data <- products_data %>%
  distinct()

customers_data <- customers_data %>%
  distinct()
```

```r
# Check quantity values

transactions_data$Quantity
```

```r
# Count invalid quantities

sum(transactions_data$Quantity <= 0)
```

```r
# Remove zero and negative quantities

transactions_data <- transactions_data %>%
  filter(Quantity > 0)
```

```r
# Check number of transactions after cleaning

nrow(transactions_data)
```

```r
# Check product prices

products_data$UnitPrice
```

```r
# Count invalid prices

sum(products_data$UnitPrice <= 0)
```

```r
# Remove zero and negative prices

products_data <- products_data %>%
  filter(UnitPrice > 0)
```

```r
# Check number of products after cleaning

nrow(products_data)
```

```r
# Combine transactions with product prices

transactions_clean <- transactions_data %>%
  left_join(
    products_data,
    by = "StockCode"
  )

# Display result
transactions_clean
```

```r
# Calculate Revenue

transactions_clean <- transactions_clean %>%
  mutate(
    Revenue = Quantity * UnitPrice
  )

# Display final cleaned transactions
transactions_clean
```

```r
# View revenue

transactions_clean %>%
  select(
    InvoiceNo,
    StockCode,
    Quantity,
    UnitPrice,
    Revenue
  )
```

```r
# Integrate transaction and customer data

final_data <- transactions_clean %>%
  left_join(
    customers_data,
    by = "CustomerID"
  )

# Display integrated dataset
final_data
```

```r
# Check dimensions of final integrated dataset

dim(final_data)
```

```r
# Number of rows and columns

nrow(final_data)
ncol(final_data)
```

```r
# Check unmatched product records

unmatched_products <- transactions_data %>%
  anti_join(
    products_data,
    by = "StockCode"
  )

unmatched_products
```

```r
nrow(unmatched_products)
```

```r
# Check unmatched customer records

unmatched_customers <- transactions_data %>%
  anti_join(
    customers_data,
    by = "CustomerID"
  )

unmatched_customers
```

```r
nrow(unmatched_customers)
```

```r
# Calculate total sales revenue

total_revenue <- final_data %>%
  summarise(
    Total_Revenue = sum(Revenue)
  )

total_revenue
```

```r
# Top 5 products based on revenue

top_products <- final_data %>%
  group_by(StockCode, Description) %>%
  summarise(
    Total_Revenue = sum(Revenue),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Revenue)) %>%
  slice_head(n = 5)

top_products
```

```r
# Top 5 countries based on revenue

top_countries <- final_data %>%
  group_by(Country) %>%
  summarise(
    Total_Revenue = sum(Revenue),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Revenue)) %>%
  slice_head(n = 5)

top_countries
```

```r
# Top 5 customers based on total purchase value

top_customers <- final_data %>%
  group_by(CustomerID, Country) %>%
  summarise(
    Total_Purchase = sum(Revenue),
    .groups = "drop"
  ) %>%
  arrange(desc(Total_Purchase)) %>%
  slice_head(n = 5)

top_customers
```

```r
# Calculate purchase value for every customer

customer_value <- final_data %>%
  group_by(CustomerID, Country) %>%
  summarise(
    Total_Purchase = sum(Revenue),
    .groups = "drop"
  )

customer_value
```

```r
# Classify customers based on purchase value

customer_value <- customer_value %>%
  mutate(
    Customer_Category = case_when(
      Total_Purchase < 3000 ~ "Low Value",
      Total_Purchase < 6000 ~ "Medium Value",
      Total_Purchase < 10000 ~ "High Value",
      Total_Purchase >= 10000 ~ "Premium"
    )
  )

# Display customer classification

customer_value
```

```r
# Create SQLite database

con <- dbConnect(
  SQLite(),
  "retail_sales.db"
)

# Check database connection
con
```

```r
# Export final integrated data to SQLite

dbWriteTable(
  con,
  "retail_sales",
  final_data,
  overwrite = TRUE
)

print("Data successfully stored in retail_sales table")
```

```r
# List tables in database

dbListTables(con)
```

```r
# Display data stored in SQLite

dbReadTable(
  con,
  "retail_sales"
)
```

```r
# SQL Query 1: Top 5 customers by revenue

query1 <- "
SELECT
    CustomerID,
    Country,
    SUM(Revenue) AS Total_Revenue
FROM retail_sales
GROUP BY CustomerID, Country
ORDER BY Total_Revenue DESC
LIMIT 5;
"

top_5_customers_sql <- dbGetQuery(
  con,
  query1
)

top_5_customers_sql
```

```r
# SQL Query 2: Total revenue by country

query2 <- "
SELECT
    Country,
    SUM(Revenue) AS Total_Revenue
FROM retail_sales
GROUP BY Country
ORDER BY Total_Revenue DESC;
"

revenue_by_country_sql <- dbGetQuery(
  con,
  query2
)

revenue_by_country_sql
```

```r
# Close SQLite connection

dbDisconnect(con)

print("SQLite database connection closed")
```
